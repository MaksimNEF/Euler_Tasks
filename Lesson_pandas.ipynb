```python
data = pd.DataFrame({
    'Отдел' :['IT','HR','IT','IT','HR'],
    'Сотрудник' :['Анна',"Борис", "Вика", "Глеб" ,"Дима"],
    "Зарплата" : [100,80,110,150,120]
})

data.groupby('Отдел')['Зарплата'].mean()
data.groupby('Отдел')['Зарплата'].agg(['mean','max','min','count'])
```

```python
data.groupby(['Отдел','Зарплата']).size()
data.groupby('Отдел').agg({
    'Зарплата':'mean',
    'Сотрудник': 'count'
})
```
